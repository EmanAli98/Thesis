```python
import duckdb
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", 200)

con = duckdb.connect("../data_processed/thesis.duckdb")

df = con.execute("""
SELECT
    cds_id,
    artist,
    title,
    cds_year,
    release_id,
    release_title,
    main_artist_name,
    artist_credit_full,
    total_tracks,
    barcode
FROM cds_to_mb_exact_matches
WHERE lower(artist) = 'kate bush'
  AND lower(title) = 'the kick inside'
ORDER BY
    CASE
        WHEN barcode IS NOT NULL AND trim(barcode) <> '' THEN 1
        ELSE 0
    END DESC,
    total_tracks,
    release_id
""").fetchdf()

print(df.to_string(index=False))
```
